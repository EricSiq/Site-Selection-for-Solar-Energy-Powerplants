```python
import numpy as np
import geopandas as gpd

def create_india_grid(
    resolution: float = 0.1,
    geojson_path: str = 'india-composite.geojson'
) -> dict:
    """
    Generates a lat-lon grid at the specified resolution for each Indian state.

    Parameters:
      resolution   Grid resolution in degrees (default 0.1°).
      geojson_url  URL or local path to the India states GeoJSON.

    Returns:
      Dict[state_name, np.ndarray of shape (N, 2)] where each row is [lat, lon].
    """
    # 1. Load state boundaries
    try:
        gdf = gpd.read_file(geojson_path)
    except Exception as e:
        raise RuntimeError(f"Failed to load state boundaries file: {e}")


    # 2. Ensure WGS84 CRS
    try:
        if gdf.crs is None:
            raise ValueError("Input GeoDataFrame has no CRS defined.")
        gdf = gdf.to_crs(epsg=4326)  # convert to lat/lon if necessary :contentReference[oaicite:5]{index=5}
    except Exception as e:
        raise RuntimeError(f"CRS error in state boundaries: {e}")

    # 3. Generate grid for each state
    grids = {}
    for idx, row in gdf.iterrows():
        # Use a state name field (may vary: 'st_nm', 'NAME', etc.)
        state = row.get('st_nm') or row.get('NAME') or f"state_{idx}"
        try:
            # Extract bounding box: minx, miny, maxx, maxy :contentReference[oaicite:6]{index=6}
            min_lon, min_lat, max_lon, max_lat = row.geometry.bounds

            # Create 1D arrays of latitudes and longitudes
            lats = np.arange(min_lat, max_lat + resolution, resolution)    # evenly spaced latitudes :contentReference[oaicite:7]{index=7}
            lons = np.arange(min_lon, max_lon + resolution, resolution)    # evenly spaced longitudes :contentReference[oaicite:8]{index=8}

            # Form meshgrid of all (lon, lat) pairs :contentReference[oaicite:9]{index=9}
            lon_grid, lat_grid = np.meshgrid(lons, lats)
            coords = np.column_stack([lat_grid.ravel(), lon_grid.ravel()])

            grids[state] = coords
        except Exception as e:
            # Warn but continue with other states :contentReference[oaicite:10]{index=10}
            print(f"Warning: failed to generate grid for {state}: {e}")

    return grids

if __name__ == "__main__":
    # Example usage
    try:
        grid_dict = create_india_grid()
        total_pts = sum(arr.shape[0] for arr in grid_dict.values())
        print(f"Generated {total_pts} total grid points across {len(grid_dict)} states.")
    except RuntimeError as err:
        print(f"Error: {err}")
```

```python
from datetime import date, datetime
import requests
import pandas as pd
from io import StringIO
from typing import Union, Optional, List

class PowerAPI:
    BASE_URL = "https://power.larc.nasa.gov/api/temporal/daily/point"

    def __init__(
        self,
        start: Union[date, datetime],
        end:   Union[date, datetime],
        lon:   float,
        lat:   float,
        parameters: Optional[List[str]] = None
    ):
        self.start = pd.to_datetime(start).strftime("%Y%m%d")
        self.end   = pd.to_datetime(end).strftime("%Y%m%d")
        self.lon   = lon
        self.lat   = lat
        # Correct solar irradiance and meteorology codes
        self.parameters = parameters or [
            "ALLSKY_SFC_SW_DWN",   # GHI
            "ALLSKY_SFC_SW_DIFF",  # DHI
            "ALLSKY_SFC_SW_DNI",   # DNI
            "T2M",                 # Air temperature
            "WS10M"                # Wind speed @10 m
        ]
        self.community = "re"    # must be lowercase

    def _build_url(self) -> str:
        params = {
            "parameters": ",".join(self.parameters),
            "community":  self.community,
            "longitude":  self.lon,
            "latitude":   self.lat,
            "start":      self.start,
            "end":        self.end,
            "format":     "CSV"
        }
        qs = "&".join(f"{k}={v}" for k, v in params.items())
        return f"{self.BASE_URL}?{qs}"

    def get_weather(self) -> pd.DataFrame:
        """
        Fetches the CSV, strips off all lines before the true header,
        and returns a DataFrame indexed by a combined date column.
        """
        # 1. Retrieve raw CSV text
        resp = requests.get(self._build_url())
        resp.raise_for_status()
        raw = resp.text.splitlines()

        # 2. Find the header line index
        header_idx = None
        for i, line in enumerate(raw):
            if line.strip().startswith("YEAR,MO,DY"):
                header_idx = i
                break
        if header_idx is None:
            raise ValueError("Could not find CSV header in NASA POWER response")

        # 3. Extract only from header onward
        csv_block = "\n".join(raw[header_idx:])

        # 4. Read CSV into DataFrame
        df = pd.read_csv(StringIO(csv_block))

        # 5. Combine YEAR/MO/DY into datetime index
        df["date"] = pd.to_datetime(
            df["YEAR"].astype(str).str.zfill(4) +
            df["MO"].astype(int).astype(str).str.zfill(2) +
            df["DY"].astype(int).astype(str).str.zfill(2),
            format="%Y%m%d"
        )
        df = df.drop(columns=["YEAR","MO","DY"]).set_index("date")
        return df

# Example usage:
if __name__ == "__main__":
    api = PowerAPI(date(2020,1,1), date(2020,12,31), lon=88.36, lat=22.57)
    df = api.get_weather()
    print(df.head())
    print("\nColumns:", df.columns.tolist())
```

```python
import numpy as np
import pandas as pd
from datetime import date
from folium.plugins import HeatMap
import folium

# --- 1. YOUR PowerAPI CLASS (assumed already imported) ---
# from your_module import PowerAPI

# --- 2. Define grid over India ---
lon_min, lon_max = 68.0, 98.0   # approximate Indian extent
lat_min, lat_max =  6.0, 38.0
resolution = 1.0                # degrees; choose coarser for speed

lons = np.arange(lon_min, lon_max, resolution)
lats = np.arange(lat_min, lat_max, resolution)
grid_points = [(lat, lon) for lat in lats for lon in lons]

# --- 3. Batch-fetch mean GHI for each point ---
results = []
for lat, lon in grid_points:
    api = PowerAPI(
        start=date(2020,1,1),
        end=  date(2020,12,31),
        lon=lon, lat=lat
    )
    df = api.get_weather()
    # extract the GHI column by its code
    ghi_series = df["ALLSKY_SFC_SW_DWN"]
    ghi_mean = ghi_series.mean()
    results.append({"lat": lat, "lon": lon, "ghi_mean": ghi_mean})

# build the DataFrame
df_results = pd.DataFrame(results)

# quick sanity check
print(df_results.head())

# --- 4. Convert to heat_data and plot ---
heat_data = df_results[["lat","lon","ghi_mean"]].values.tolist()

m = folium.Map(location=[22,79], zoom_start=5)
HeatMap(
    data=heat_data,
    radius=8,
    blur=15,
    max_zoom=10
).add_to(m)

# in Jupyter this will render automatically:
m
# or save to HTML:
#m.save("india_ghi_heatmap.html")
```
